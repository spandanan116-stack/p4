```python
#Read N numbers from the console and create a list. Develop a program to print mean, variance and standard deviation with suitable messages.
x = int(input("enter n number:"))
l = []
for i in range(x):
    l.append(int(input()))
print(l)
mean = sum(l)/x
v = 0
for i in l:
    v = v + (i-mean)**2
var = v/x
import math
sd = math.sqrt(var)
print("Mean={mean:0.2f},Variance={var:0.2f},Standard deviation={sd:0.2f}")
```

```python
#Read a multi-digit number(as chars) from the console. Develop a program to print the frequency of each digit with a suitable message
n = input("enter a multi-digit number:")
freq = {}
for digit in n:
    if digit.isdigit():
        freq[digit] = freq.get(digit,0) + 1
print("\nDigit frequencies:")
for digit in sorted(freq):
    print(f"Digit {digit} occurs {freq[digit]} time(s)")
    
        
```
